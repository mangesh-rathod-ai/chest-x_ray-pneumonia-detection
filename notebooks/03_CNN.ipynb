# 03 - Baseline CNN (from scratch)

A small 4-block CNN trained without any pretrained weights. It is our reference point: transfer learning (notebook 04) should beat it.

```python
import sys, warnings
from pathlib import Path
warnings.filterwarnings("ignore")

# Make the project's `src` package importable from inside notebooks/
ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src" / "config.py").exists()), Path.cwd())
sys.path.insert(0, str(ROOT))

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

from src import config
from src.utils import find_dataset_root, count_images_per_class

# Point this at data/dummy to try the notebook without the real dataset
DATA_DIR = config.DATA_DIR
ROOT_DATA = find_dataset_root(DATA_DIR)
print("Dataset root:", ROOT_DATA)

from src import train, models
from IPython.display import Image as IPImage, display
```

```python
EPOCHS = 15
BATCH_SIZE = 32
IMG_SIZE = 224

model = models.build_model("baseline_cnn", IMG_SIZE, weights=None, freeze_backbone=False)
model.summary(line_length=100)
```

## Train

This calls the same `src.train` code as the command line. Outputs go to `models/baseline_cnn.keras` and `outputs/baseline/`, so the final model in `models/medical_model.keras` is not overwritten.

```python
train.main([
    "--data-dir", str(DATA_DIR),
    "--backbone", "baseline_cnn",
    "--epochs", str(EPOCHS),
    "--batch-size", str(BATCH_SIZE),
    "--img-size", str(IMG_SIZE),
    "--model-path", str(config.MODELS_DIR / "baseline_cnn.keras"),
    "--config-path", str(config.MODELS_DIR / "baseline_cnn_config.json"),
    "--output-dir", str(config.OUTPUTS_DIR / "baseline"),
])
```
